# Lokale LLMs mit Python nutzen

```python
MODEL = "peterparker" # Name meines Modells wie ihn OLLAMA kennt
```

```python
from ollama import chat, generate
```

## Chatten mit dem Modell

```python
messages = [{
    "role": "User",
    "content": "Hallo ich bin Michael, wer bist du?"
}]

response = chat(MODEL, messages)

print(response["message"]["content"])
```

## One-Shot Generator Kommunikation mit LLM

```python
from os import listdir

for file in listdir("quotes"):

    with open("quotes/"+file, "r") as f:
        content = f.read()

    prompt = "Interpretiere dieses Zitat: " + content

    response = generate(MODEL, prompt)

    with open("output/"+file, "w") as f:
        f.write(response["response"])
```

## Mit Google AI Studio

```python
from google import genai
from os import environ

API_KEY = environ["API_KEY"]

client = genai.Client(api_key=API_KEY)

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="Explain how AI works in detail",
)

print(response.text)
```
